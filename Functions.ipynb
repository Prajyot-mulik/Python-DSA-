# Function examples

```python
def tmp_converter(temp, unit):
    if unit == 'c':
        return temp * 9/5 + 32
    elif unit == 'f':
        return (temp - 32) * 5/9
    else:
        return "Invalid unit"
    

print(tmp_converter(25, 'c'))
    
```

* Lambda function are small anonymous fucntion defined using the Lambda keyword. They can have any number of argument but only one expression. They are commonly used for short opretion or as argument to higher-order funtion

```python
lambda argument: expression
```

```python
even = lambda num:num % 2 ==0 

even(34)
```

```python

## map() - applies a function to all items in a list
def square(number):
    return number**2

square(6)
```
