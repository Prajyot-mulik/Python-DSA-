1) You are given an integer n. your tak is to return a square pattern of size nxn made up of the charater '*' represented as a list of string 

```python
def generate_squares(n):
    squre = []
    for i in range(n):
        squre.append(n * '*')
    return squre

generate_squares(3)
```

```python
def generate_holly(n):
    holly = []
    for i in range(n):
        if i == 0 or i == n-1:
            holly.append(n * '*')
        else:
            holly.append('*' + ' ' * (n-2) + '*')
    return holly

generate_holly(3)
```
