1) You are given an integer n. your tak is to return a square pattern of size nxn made up of the charater '*' represented as a list of string 

```python
def generate_squares(n):
    squre = []
    for i in range(n):
        squre.append(n * '*')
    return squre

generate_squares(3)
```

```python
def generate_holly(n):
    holly = []
    for i in range(n):
        if i == 0 or i == n-1:
            holly.append(n * '*')
        else:
            holly.append('*' + ' ' * (n-2) + '*')
    return holly

generate_holly(3)
```

```python

# daimand pattern
def daimand(n):
    list = []
    # upper part
    for i in range(1, n+1):
        stars = ( 2 * i - 1)
        spaces = (n - i)
        list.append(' ' * spaces + '*' * stars + ' ' * spaces)

    
        

    for i in range(n - 1, 0 , -1):
        stars = ( 2 * i - 1)
        spaces = (n - i)
        list.append(' ' * spaces + '*' * stars + ' ' * spaces)
        
    return list
        

daimand(5)
  

```

```python
#right angle triangle
def right_angle_triangle(n):
    result = []
    for i in range(1, n+1):
        spaces = ' ' * (n - i)
        stars = '*' * i
        result.append(spaces + stars)
    return result

right_angle_triangle(5)
```

```python
 # sand box pattern
def sandbox(n):
    result = []
    for i in range(n, 0 , -1):
        space = n - i 
        stars = 2*i - 1
        result.append(' ' * space + '*' * stars + ' ' * space)
    for i in range(2, n+1):
        space = n - i
        start = 2*i - 1
        result.append(' ' * space + '*' * start + ' ' * space)
    return result

sandbox(5)
    
```
